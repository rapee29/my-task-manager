Part 1: Single Number Table

```python
n = int(input("Enter a number for its multiplication table (e.g., 7): "))

for i in range(1, 13):
    print(f"{n} x {i} = {n * i}")

print()
```

Part 2: Full Multiplication Grid (12 × 12)

```python
print("12 x 12 Multiplication Grid\n")

for i in range(1, 13):
    for j in range(1, 13):
        print(f"{i * j:4d}", end="")
    print()
```
