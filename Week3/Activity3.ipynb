# Activity 3.1: Debugging Infinite Loops & Loop Logic Errors

A: Infinite Loop

```python
# โค้ดที่แก้แล้ว

counter = 0

while counter < 5:
    print(counter)
    counter += 1
```

แก้อะไรบ้าง

เพิ่ม counter += 1 ภายใน while loop
เดิมไม่มีการเพิ่มค่า counter ทำให้ counter เป็น 0 ตลอด และลูปทำงานไม่สิ้นสุด (Infinite Loop)
หลังแก้ counter จะเพิ่มทีละ 1 จนถึง 5 แล้วลูปจะหยุด

ข้อ B: Incorrect Range

```python
# โค้ดที่แก้แล้ว

for i in range(1, 6):
    print(i)
```

แก้อะไรบ้าง

เปลี่ยน range(1, 5) เป็น range(1, 6)
เพราะ range() จะไม่รวมค่าตัวสุดท้าย (stop)
ถ้าต้องการแสดงเลข 1 ถึง 5 ต้องกำหนดค่า stop เป็น 6

C: Misplaced Break

```python
# โค้ดที่แก้แล้ว

for char in "hello":
    print(char)
    if char == 'l':
        print("Found 'l', stopping!")
        break
```

แก้อะไรบ้าง

ย้าย print(char) ไปไว้ก่อน break
เดิมเมื่อเจอ 'l' โปรแกรมจะ break ทันที ทำให้ตัวอักษร 'l' ไม่ถูกแสดง
หลังแก้ โปรแกรมจะแสดง 'l' ก่อน แล้วจึงแสดงข้อความ "Found 'l', stopping!" และออกจากลูป

# Activity 3.2: Pattern Printing with Nested Loops

1: Square of Asterisks

```python
for i in range(5):
    for j in range(5):
        print("*", end="")
    print()
```

2: Right-Angled Triangle

```python
for i in range(1, 6):
    for j in range(i):
        print("*", end="")
    print()
```

Challenge 1: Inverted Right-Angled Triangle

```python
for i in range(5, 0, -1):
    for j in range(i):
        print("*", end="")
    print()
```

Challenge 2: Pyramid Pattern

```python
for i in range(1, 6):
    for j in range(5 - i):
        print(" ", end="")

    for j in range(2 * i - 1):
        print("*", end="")

    print()
```
