ทำไมถึงเป็น type <class 'float'>: ในโค้ดบรรทัด x = 10 ตัวแปร x เป็นตัวเลขจำนวนเต็ม (Integer หรือ int) ส่วน y = 3.5 เป็นตัวเลขทศนิยม (Float)
เมื่อคุณนำ x และ y มาบวกกัน (z = x + y) Python จะทำการแปลงผลลัพธ์ให้เป็นทศนิยม (Float) โดยอัตโนมัติ เพื่อป้องกันไม่ให้ข้อมูลส่วนที่เป็นทศนิยมสูญหายไปครับ ดังนั้น 10 + 3.5 จึงเท่ากับ 13.5 และมีประเภทข้อมูลเป็น float

````python
# Puzzler 2
name = "Alice"
age = "30" # Note: This is a string
message = name + " is " + age + " years old."
print(message)
print(type(age))
````


* **ได้ผลลัพธ์:** `Alice is 30 years old.`
* **ทำไมถึงเป็น type `<class 'str'>`:**
หากสังเกตที่บรรทัด `age = "30"` จะเห็นว่าตัวเลข 30 ถูกเขียนอยู่ภายใต้เครื่องหมายคำพูด (Double quotes `""`)
ในภาษา Python **ข้อมูลใดๆ ก็ตามที่ถูกครอบด้วยเครื่องหมายคำพูด จะถือว่าเป็นข้อความ (String หรือ `str`) ทันที** แม้ว่าข้างในนั้นจะเป็นตัวเลขก็ตาม มันจะถูกมองเป็นแค่ตัวอักษร '3' กับ '0' เรียงต่อกัน ไม่สามารถนำไปบวกหรือคำนวณทางคณิตศาสตร์ได้ครับ ดังนั้นเมื่อเช็ค `type(age)` จึงได้ผลลัพธ์ออกมาเป็น `<class 'str'>`

````python
# Puzzler 3
is_sunny = True
is_raining = False
print(is_sunny and is_raining)
print(type(is_sunny))
````

ได้ผลลัพธ์: False และ <class 'bool'>

ทำไมถึงได้ผลลัพธ์นี้: 1. is_sunny and is_raining ได้ False: เพราะตัวดำเนินการ and (และ) มีเงื่อนไขว่า "ผลลัพธ์จะเป็นจริง (True) ได้เพียงกรณีเดียวเท่านั้น คือข้อมูลทั้งสองฝั่งต้องเป็น True ทั้งคู่" แต่ในโค้ดนี้ is_sunny เป็น True และ is_raining เป็น False เมื่อเจอ and เข้าไป ผลลัพธ์รวมจึงออกมาเป็น False ครับ
2. ทำไมถึงเป็น type <class 'bool'>: ตัวแปรที่มีค่าเป็น True หรือ False (ขึ้นต้นด้วยตัวพิมพ์ใหญ่ใน Python) จัดอยู่ในประเภทข้อมูลที่เรียกว่า Boolean (ตรรกศาสตร์) ตัว Python จึงย่อประเภทนี้ให้เราเห็นเป็น <class 'bool'>

````python
# Grabs the input and immediately converts it to an integer
num_int = int(input("Enter a number: "))

result = num_int + 5
print(result)
````
