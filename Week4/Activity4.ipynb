Activity 4.1: List vs. Tuple: Scenario Matching

สถานการณ์	ใช้ List หรือ Tuple	เหตุผล
1. A collection of daily temperatures recorded over a week	List	อุณหภูมิสามารถเพิ่ม แก้ไข หรือลบข้อมูลได้
2. The RGB color values for a specific shade (255, 0, 0)	Tuple	ค่าสี RGB เป็นค่าคงที่ ไม่ควรแก้ไข (Immutable)
3. Items in a shopping cart that can be added or removed	List	สินค้าในตะกร้าสามารถเพิ่มและลบได้
4. The days of the week in order	Tuple	ลำดับวันในสัปดาห์เป็นข้อมูลคงที่ ไม่เปลี่ยนแปลง
5. Student scores on a series of assignments, which might be updated	List	คะแนนสามารถแก้ไขได้เมื่อมีการอัปเดต
6. GPS coordinates for a fixed landmark	Tuple	พิกัดของสถานที่หนึ่งเป็นค่าคงที่ ไม่ควรเปลี่ยน
สรุปหลักการ
List ([]) → ใช้กับข้อมูลที่ แก้ไขได้ (Mutable) เช่น เพิ่ม ลบ หรือแก้ไขข้อมูล
Tuple (()) → ใช้กับข้อมูลที่ ไม่ควรเปลี่ยนแปลง (Immutable) เช่น พิกัด สี RGB หรือค่าคงที่ต่าง ๆ

Activity 4.2:Interactive To-Do List

```python
# Interactive To-Do List

todo_list = []

while True:
    print("\n===== TO-DO LIST =====")
    print("1. Add a task")
    print("2. View tasks")
    print("3. Mark task as complete")
    print("4. Exit")

    choice = input("Enter your choice (1-4): ")

    if choice == "1":
        task = input("Enter a new task: ")
        todo_list.append(task)
        print("Task added successfully!")

    elif choice == "2":
        if len(todo_list) == 0:
            print("No tasks available.")
        else:
            print("\nYour Tasks:")
            for i in range(len(todo_list)):
                print(f"{i + 1}. {todo_list[i]}")

    elif choice == "3":
        if len(todo_list) == 0:
            print("No tasks to remove.")
        else:
            print("\nYour Tasks:")
            for i in range(len(todo_list)):
                print(f"{i + 1}. {todo_list[i]}")

            remove_choice = input("Remove by (1) Number or (2) Name? ")

            if remove_choice == "1":
                try:
                    index = int(input("Enter task number: "))
                    if 1 <= index <= len(todo_list):
                        completed = todo_list.pop(index - 1)
                        print(f'"{completed}" completed and removed.')
                    else:
                        print("Invalid task number.")
                except ValueError:
                    print("Please enter a valid number.")

            elif remove_choice == "2":
                name = input("Enter task name: ")
                if name in todo_list:
                    todo_list.remove(name)
                    print(f'"{name}" completed and removed.')
                else:
                    print("Task not found.")

            else:
                print("Invalid choice.")

    elif choice == "4":
        print("Goodbye!")
        break

    else:
        print("Invalid choice. Please try again.")
```
